```python
# Forecast of orders for next 14 days 
# in order to make file clean, manual modifciations done to the "cleaned_monthly_orders.csv" and saved as "cleaned_monthly_orders_SJ.csv" 
from statsmodels.tsa.holtwinters import Holt
import numpy as np

# Load the uploaded dataset
file_path = "C:/Users/suneth.jayamanne/.cache/kagglehub/datasets/oscarm524/demand-forecast-for-optimized-inventory-planning/versions/5/cleaned_monthly_orders_SJ.csv"
df = pd.read_csv(file_path)

# Display basic information and first few rows
df.info(), df.head()

# Prepare forecast dictionary
forecast_results = {}

# Extract only the monthly columns
monthly_columns = ['Month 1', 'Month 2', 'Month 3', 'Month 4', 'Month 5', 'Month 6']
df_monthly = df.set_index('itemID')[monthly_columns]

# Forecasting for the next 14 days (half a month approximation)
forecast_periods = 1

# Apply forecasting model for each item
for item in df_monthly.index:
    try:
        ts_data = df_monthly.loc[item].dropna()
        
        if len(ts_data) >= 3:  # Ensure at least 3 months of data
            model = Holt(ts_data)
            fit = model.fit()
            forecast = fit.forecast(forecast_periods)
            forecast_results[item] = forecast.values[0]  # Store single forecasted value
        else:
            forecast_results[item] = np.nan  # Not enough data for reliable forecasting
    except:
        forecast_results[item] = np.nan  # Handle errors gracefully

# Convert forecast results into a DataFrame
forecast_df = pd.DataFrame(list(forecast_results.items()), columns=["itemID", "Forecasted Orders (14 Days)"])

# Handle NaN values before applying transformation
forecast_df["Forecasted Orders (14 Days)"] = forecast_df["Forecasted Orders (14 Days)"].fillna(0)

# Convert negative values to zero and round up positive values to whole numbers
forecast_df["Forecasted Orders (14 Days)"] = forecast_df["Forecasted Orders (14 Days)"].apply(
    lambda x: max(0, round(x))
)

# Save and display forecast results
forecast_file_path = "C:/Users/suneth.jayamanne/.cache/kagglehub/datasets/oscarm524/demand-forecast-for-optimized-inventory-planning/versions/5/forecast_14_days.csv"
forecast_df.to_csv(forecast_file_path, index=False)

# Provide the file for download
forecast_file_path
```
